# Starter cohort analysis

Pull starter cancellations by signup month; render cohort plot.

```python
import duckdb, pandas as pd
df = duckdb.sql("""
  select signup_month, days_to_cancel, count(*) as n
  from read_csv_auto('customers.csv')
  where plan = 'starter'
  group by 1, 2
""").df()
df.head()
```

```python
cohort = df.pivot('signup_month', 'days_to_cancel', 'n').fillna(0)
ax = cohort.plot.area(stacked=True, figsize=(8, 3))
ax.set_ylabel('starter cancellations'); ax.set_xlabel('signup month')
```

```python
day30_retention = 1 - df.query('days_to_cancel <= 30').n.sum() / df.n.sum()
print(f'day-30 retention: {day30_retention:.3f}')
```
